**Exercise 1.** `top_item` should return the item with the highest TOTAL quantity, but it returns the item on the single largest line. Fix it.

```python
def top_item(lines):
    best_item, best_qty = None, 0
    for item, qty in lines:
        if qty > best_qty:
            best_item, best_qty = item, qty
    return best_item
```
